# Ferramentas matemáticas úteis
Antes de começar a discussão sobre as equações de Maxwell, é útil introduzir o 
operador $\nabla$ (nabla), o conceito de fluxo e circulação de um campo vetorial, e dois teoremas importantes que serão utilizados no decorrer do texto: o teorema da divergência (Gauss) e do rotacional (Stokes).

## Fluxo e Circulação

### Fluxo
Seja $\mathbf F$ um campo vetorial definido numa região do espaço. Seja $S$ uma superfície orientável 
contida nessa região. Por ser orientável, podemos associar a cada ponto em $S$ um vetor unitário $\mathbf n$, que define a normal externa da superfície. O fluxo de $\mathbf F$ sobre $S$ é definido por 
$$
\Phi_{\mathbf F}(S) \equiv \iint_S \mathbf F \cdot d\mathbf A \quad , 
$$
onde $d\mathbf A \equiv \mathbf n dA$ é um elemento de área orientado.

### Circulação
Seja $\mathbf F$ um campo vetorial definido numa região do espaço. Seja $C$ uma curva fechada contida nessa região e parametrizada por $\mathbf l(t)$. A circulação de $\mathbf F$ sobre $C$ é dada por
$$
\Gamma_\mathbf F (C) \equiv \int_C \mathbf F \cdot d\mathbf l \quad ,
$$
onde $d\mathbf l = (d\mathbf l /dt) dt$.

## Operador Nabla 
$\nabla$ é um operador que age em funções diferenciáveis cujo domínio são vetores em $\mathbb{R}^n$. Em coordenadas cartesianas $(x^1, \dots , x^n)$ e base natural $\{\mathbf{e}_1, \dots, \mathbf{e}_n\}$, podemos considerá-lo como sendo um "vetor de operadores derivada":  $\nabla \equiv \mathbf{e}_1\partial_1 + \dots + \mathbf{e}_n\partial_n$, onde $\partial_i f \equiv \partial f/\partial x_i$. Por si só, essa expressão não tem significado, e sua interpretação depende da natureza função.

### Gradiente
Se $F$ é um campo escalar, definimos o gradiente de $F$ como sendo 
$$
\text{grad }F \equiv \mathbf{e}_1 \partial_1 F + \dots + \mathbf{e}_n \partial_n F = \nabla F \quad .  
$$
Considere um deslocamento infinitesimal $d\mathbf{x} = \mathbf{u}dt$, onde $\mathbf{u}$ é um vetor unitário e $t$ é um parâmetro, de modo que $dx^i = u^i dt$. A variação $dF$ é dada por
\begin{equation*}
F(\mathbf{x}+ d\mathbf{x}) - F(\mathbf{x}) \equiv dF = \partial_i F \, dx^i = dt(\mathbf u \cdot \nabla F) \quad .
\end{equation*}
Portanto, $dF$ é máximo quando $\nabla F // \mathbf u$, e zero quando $\nabla F \perp \mathbf u$. i.e., $\nabla F$ aponta na direção de maior taxa de variação de $F$, e é normal à superfícies $F =$ constante. Na equação acima e no resto desse trabalho, foi utilizada a convenção de notação de Eistein, onde índices repetidos em posições diferentes (em cima e em baixo) significam que há um somatório implícito sobre esse índice.

### Divergência 
Seja $\mathbf F$ um campo vetorial definida em uma região do espaço. Seja V uma região contida no domínio de $\mathbf F$. Seja $|V|$ o volume de $V$, e $\partial V$ sua fronteira. Dado um ponto $\mathbf x$ no domínio de $\mathbf F$, tomamos uma região fechada $V \ni \mathbf x$, e calculamos a razão entre $\Phi_\mathbf F(\partial V)$ e $|V|$; fazendo o volume ir a zero, o valor limite é a divergência de $F$ em $\mathbf x$. Matematicamente, temos
\begin{equation}
\text{div }\mathbf{F} \equiv \lim\limits_{|V| \to 0} \frac{\Phi_F(\partial V)}{|V|} \quad ,
\end{equation}

Em coordenadas cartesianas, a expressão para a divergência de um campo vetorial é $\text{div } \mathbf F = \partial_i F^i = \nabla \cdot \mathbf{F}$. 

Se $\nabla \cdot \mathbf{F} > 0$, então $\mathbf{x}$ é uma fonte de $\mathbf{F}$; se $\nabla \cdot \mathbf{F} < 0$, então $\mathbf{x}$ é um sorvedouro de $\mathbf{F}$. 

### Rotacional 
Seja $\mathbf F$ um campo vetorial. Seja $A$ uma superfície orientável contida no domínio de $\mathbf F$. Analogamente ao caso anterior, denotamos por $|A|$ sua área e por $\partial A$ sua fronteira (uma curva fechada). A componente de $\text{rot }\mathbf{F}$ na direção de um vetor unitário $\mathbf{n}$ num ponto $\mathbf{x}$ é definida como
\begin{equation}
\mathbf{n} \cdot (\text{rot }\mathbf{F}) \equiv \lim\limits_{|A| \to 0} \frac{\Gamma_\mathbf F (\partial A)}{|A|} \quad ,
\end{equation}
onde escolhemos as superfícies $A \ni \mathbf x$ de modo que $A \perp \mathbf n$ em $\mathbf x$, $\forall A$. 
Em coordenadas cartesianas, a fórmula para o rotacional é 
\begin{equation}
\text{rot }\mathbf F = \epsilon^{ijk}\partial_i F_j \mathbf{e}_k = \nabla \times \mathbf{F} \quad,
\end{equation}
onde $\epsilon^{ijk}$ é o tensor de Levi-Civita. O rotacional, diferentemente das outras aplicações do operador $\nabla$, requer que $\mathbf{F}$ seja uma função $\mathbb{R}^3 \to \mathbb{R}^3$, sendo $\nabla \times \mathbf{F}$ também uma função desse tipo. 

### Laplaciano
Para um campo escalar $F$, o laplaciano é definido como 
$$
\text{div }(\text{grad }F) = \nabla \cdot \nabla F \equiv \nabla^2 F = \partial_1^2 F + \dots + \partial_n^2 F \quad.
$$

Enquanto que para um campo vetorial $\mathbf F$, 
$$
\nabla^2 \mathbf F = \nabla^2 (F^i \mathbf{e}_i) = (\nabla^2 F^i) \mathbf{e}_i \quad .
$$

As seguinte identidades serão úteis:
\begin{equation}
    \nabla \cdot \nabla \times \mathbf F = 0 \quad,
\end{equation}
\begin{equation}
    \nabla \times \nabla \times \mathbf F = \nabla(\nabla \cdot \mathbf F) - \nabla^2 \mathbf F \quad.
\end{equation}


## Teorema da divergência (de Gauss)
**Teorema da divergência**. 
Seja $\mathbf F$ um campo vetorial continuamente diferenciável. Seja $V$ um subconjunto de $\mathbb{R}^n$ contido no domínio de $\mathbf F$, e delimitado pela superfície orientável $\partial V$. Temos então
\begin{equation}
\iint_{\partial V} \mathbf{F} \cdot d\mathbf{A} = \iiint_V \nabla \cdot \mathbf{F} dV \quad .
\end{equation}

_Prova (informal)._ Particionamos $V$ em vários subconjuntos de volume $\Delta V_i$, delimitados cada um por uma superfície orientada $S_i$. No cálculo sobre o fluxo total de $\mathbf F$ sobre todas as superfícies $S_i$, se dois elementos de volume compartilharem um mesmo elemento de superfície, suas contribuições serão opostas, devido a suas orientações. Assim, os únicos termos diferentes de zero serão os da fronteira de $V$, e temos 
\begin{align}
\Phi_F(\partial V) &= \sum_i \Phi_F(S_i) \notag \\
\implies \iint_{\partial V} \mathbf{F} \cdot d\mathbf{A} &= \sum_i \iint_{S_i} \mathbf{F} \cdot d\mathbf{A} \notag \\
&= \sum_i \left( \frac{1}{\Delta V_i} \iint_{S_i} \mathbf{F} \cdot d\mathbf{A} \right) \Delta V_i \quad .
\end{align}

O argumento pode ser feito para $\Delta V_i$ arbitrariamente pequenos. No limite em que $\Delta V_i \to 0$, o termo entre parênteses se torna a div $\mathbf{F}$, e o somatório se torna uma integral sobre o volume $V$. $\quad \square$

## Teorema do rotacional (de Stokes)

**Teorema do rotacional.** Seja $S$ uma superfície orientável delimitada pelo contorno $\partial S$. Se $\mathbf{F}$ é um campo vetorial 
continuamente diferenciável, então
\begin{equation}
\iint_S \nabla \times \mathbf{F} \cdot d\mathbf{A} = \oint_{\partial S} \mathbf{F} \cdot d\mathbf{l} \quad .
\end{equation}

A prova deste será omitida, mas segue a mesma lógica da prova anterior.
