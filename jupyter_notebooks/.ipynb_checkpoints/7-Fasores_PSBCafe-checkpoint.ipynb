# Fasores

Em muitos casos na física lidamos com sistemas que evoluem como uma função senoidal.
Nesses casos, é útil pensar na função como sendo uma das coordenadas de um vetor que rotaciona num plano,
com frequência angular constante. 
Essa ferramenta matemática recebe o nome de fasor. 

## Parametrização de uma circunferência

Considere uma circunferência de raio $R$ centrada na origem de um plano cartesiano. 
Se $P=(x,y)$ é um ponto que pertence à circunferência, então, por definição, 
$$
x = R\cos\theta , \qquad y = R\sin\theta \;,
$$
onde $\theta$ é o angulo entre o eixo $x$ positivo e o raio que passa por $P$, orientado no sentido anti-horário.
Fazendo $\theta$ variar de $0$ a $2\pi$, temos que o ponto percorre toda a circunferência uma vez, com velocidade angular $\dot \theta$. 


## Representação complexa

Quando lidamos com coordenadas polares, muitas vezes se mostra útil trabalhar com números complexos. 
O plano complexo é a soma direta do espaço $\mathbb R$ com $\mathbb I$. 
i.e., todo número complexo $z$ pode ser escrito na forma $x+iy$, onde nomeamos $x$ como sendo sua parte real $\text{Re}(z)$, 
e $y$ como sendo sua parte imaginária $\text{Im}(z)$.
Também definimos a norma de algum $z$ como sendo $|z|$, tal que $|z|^2 = z \bar z = (x+iy)(x-iy) = x^2 + y^2$. 
Geometricamente, $|z|$ é o tamanho do segmento de reta ligando a origem a $z$.

Mediante inspeção da representação geométrica de algum $z$ no plano complexo, é possível determinar que, $\forall z \in \mathbb C$, temos
$$
\text{Re}(z) = |z|\cos\theta \;, \qquad \text{Im}(z) = |z|\sin\theta \;
$$
onde $\theta$ é o ângulo entre o eixo real positivo e o segmento de reta que liga a origem a $z$. 
Assim, qualquer $z \in \mathbb C$ pode ser representado em sua forma polar:
$$
z = \text{Re}(z) + i \, \text{Im}(z) = |z|(\cos\theta + i \sin\theta) = |z|e^{i\theta} \;,
$$
onde fizemos uso da equação de Euler, a ser demonstrada a seguir. 

Com a representação complexa acima, podemos estudar a evolução de qualquer fenômeno ondulatório com sendo uma função complexa
em sua representação polar.
Assim os processos de superposição, e diferenciação das soluções se tornam mais simples.
Para retornar ao caso físico, onde a função é senoidal, basta considerar a projeção da função complexa nos eixos real ou imaginário.



### Demonstração da eq. de Euler

#### Método 1 (por série de Taylor)
Considere a expansão de $e^x$ em torno de $x=0$:
$$
e^x = \sum\limits_{n=0}^\infty \frac{x^n}{n!} \;.
$$

Fazendo $x = i \theta$, obtemos
$$
e^{i\theta} = \sum\limits_{n=0}^\infty \frac{(i\theta)^n}{n!} \;.
$$

Separamos as potências pares das ímpares:
$$
e^{i\theta} = \sum\limits_{n=0}^\infty \frac{(i\theta)^{2n}}{(2n)!} +
\sum\limits_{n=0}^\infty \frac{(i\theta)^{2n+1}}{(2n+1)!} \;.
$$

Temos que 
$$
i^{2n} = (i^2)^n = (-1)^n \;, \qquad  i^{2n+1} = i\,i^{2n} = i\,(-1)^n \;.
$$

Substituindo na expansão, obtemos
$$
e^{i\theta} = \sum\limits_{n=0}^\infty \frac{(-1)^n\theta^{2n}}{(2n)!} +
i\sum\limits_{n=0}^\infty \frac{(-1)^n\theta^{2n+1}}{(2n+1)!} \;.
$$

Mas os somatórios acima são exatamente as expansões de $\cos\theta$ e $\sin\theta$, respectivamente, 
em série de Taylor ao redor de $0$. 
Assim, 
$$
e^{i\theta} = \cos\theta + i \sin\theta \;. \qquad \square
$$

#### Método 2 (por EDO)
Seja $e^{i\theta}$ uma função complexa que buscamos determinar. 
Podemos representá-la como $f(\theta) + i \, g(\theta)$, onde $f$ e $g$ são funções reais.
Queremos então encontrar $f(\theta)$ e $g(\theta)$. 
Diferenciando $e^{i\theta}$ duas vezes, obtemos
\begin{gather*}
e^{i\theta} = f(\theta) + i \, g(\theta) \\
\implies i e^{i\theta} = f'(\theta) + i \, g'(\theta) \\
\implies - e^{i\theta} = f''(\theta) + i \, g''(\theta) \;.
\end{gather*}

Ou seja, 
$$
f''(\theta) + i \, g''(\theta) = -f(\theta) - i \, g(\theta) \;.
$$

Mas isso implica em 
$$\begin{cases}
f''(\theta) = -f(\theta) \\
g''(\theta) y = -g(\theta)
\end{cases}
\implies
\begin{cases}
f(\theta) = A \sin\theta + B \cos\theta \\
g(\theta) = C \sin\theta + D \cos\theta 
\end{cases} \quad ,$$
onde $A$, $B$, $C$, e $D$ são constantes a serem determinadas. 
Substituindo essas expressões para $f(\theta)$ e $g(\theta)$ em $e^{i\theta}$ e em sua derivada, 
e impondo que $e^{0}=1$, obtemos que $A=D=1$, e $B=C=0$.

Assim, temos que 
$$
e^{i\theta} = \cos\theta + i \sin\theta \;. \qquad \square
$$

## Aplicação de fasores em difração simples

### Introdução

Sabemos que a luz é uma onda eletromagnética que se propaga com velocidade $c$.
Podemos identificar a onda univocamente pelo seu campo elétrico $\mathbf E$.
Uma solução das equações de Maxwell assume a forma de onda plana, com um modo único de vibração: 
$$
\mathbf E = \mathbf{\hat e} \, \psi(\mathbf r, t)  \;, \qquad \psi(\mathbf r, t) = A \,e^{i(\mathbf k \cdot \mathbf r - \omega t)}
$$
onde $\mathbf{\hat e}$ é um vetor de polarização da luz.

Consideramos que a onda incide sobre uma pequena fenda de abertura $a$. 
Pelo princípio de Huygens-Fresnel, cada ponto da frente de onda age como uma fonte de uma 
onda esférica. 
Assim, após atravessar a fenda, a frente de onda muda de forma, e passa a se propagar 
esfericamente. 
Esse fenômeno é chamado de difração.

Consideramos que a uma distância $D$ da fenda se encontra um anteparo. 
Um ponto $P$ no anteparo recebe luz vindo de toda a extensão da fenda. 
Devido ao princípio de superposição de ondas, o valor resultante de $\mathbf E$ em $P$
será a soma dos $\mathbf E$ vindos de cada posição da fenda. 
Devido a diferença de caminho entre cada feixe, a fase de cada um deles será diferente. 
Consideramos então a possibilidade de que existam posições no anteparo em que as fases se cancelam,
gerando uma interferência destrutiva. 

Com efeito, realizando o experimento, observamos um padrão de franjas claras e escuras que se formam. 
Buscamos então determinar a posição das franjas escuras no padrão formado, e vamos utilizar do artifício dos fasores para tal.

### Calculo da superposição

```python
import matplotlib.pyplot as plt
import numpy as np

# Configuração da figura: Aumentado o tamanho horizontal (15) e reduzido o vertical (5)
fig, ax = plt.subplots(figsize=(15, 5))

# Parâmetros Geométricos Ajustados para enfatizar o regime de Fraunhofer (D >> a)
D = 14        # Distância horizontal ampliada até o anteparo
a = 1.6       # Abertura da fenda reduzida para o efeito de campo distante
P_y = 1.8      # Posição vertical do ponto P (mais baixa para raios quase paralelos)

# Redução dos limites verticais do esquema para achatar o gráfico
y_max_esquema = 3.5
y_min_esquema = -3.5

# --- 1. Desenho da Fenda com Bordas (Ocupando a mesma altura do anteparo) ---
# Borda superior da fenda
ax.plot([0, 0], [y_max_esquema, a/2], color='black', lw=4)
# Borda inferior da fenda
ax.plot([0, 0], [-a/2, y_min_esquema], color='black', lw=4)

# Indicação da largura 'a' com setas
ax.annotate('', xy=(-0.4, a/2), xytext=(-0.4, -a/2),
            arrowprops=dict(arrowstyle='<->', color='black', lw=1.2))
ax.text(-0.6, 0, '$a$', fontsize=14, ha='right', va='center')

# --- 2. Desenho do Anteparo (Mesma altura das bordas da fenda) ---
ax.plot([D, D], [y_min_esquema, y_max_esquema], color='black', lw=2.5)

# Texto "Anteparo" na HORIZONTAL
ax.text(D + 0.3, -2.0, 'Anteparo', fontsize=12, ha='left', va='center', rotation=0)

# --- 3. Elementos Geométricos Centrais ---
# Ponto O (Centro da fenda) e Ponto P (No anteparo)
ax.plot(D, P_y, 'ro', markersize=8, zorder=5) # Ponto P vermelho
ax.text(D + 0.3, P_y, '$P$', fontsize=13, va='center', ha='left')

# Linha horizontal tracejada de referência (partindo de O)
ax.plot([0, D], [0, 0], color='gray', linestyle='--', lw=1.5)

# Reta de referência que liga o centro O ao ponto P (para definir theta)
ax.plot([0, D], [0, P_y], color='red', linestyle='-.', lw=1)

# --- 4. Raios de Luz (Difração) ---
# Gerando 5 pontos equidistantes na fenda (do topo ao fundo da abertura)
y_fenda = np.linspace(a/2, -a/2, 5)

# Traçando as 5 linhas de mesma cor (azul) partindo da fenda até P
for y in y_fenda:
    ax.plot([0, D], [y, P_y], color='royalblue', alpha=0.7, lw=1.5)

# --- 5. Indicação do Ângulo Theta (\theta) ---
theta = np.arctan(P_y / D)
arc_theta = np.linspace(0, theta, 100)
arc_radius = 2.2  # Aumentado o raio do arco para melhor visualização na proporção longa
ax.plot(arc_radius * np.cos(arc_theta), arc_radius * np.sin(arc_theta), color='black', lw=1.2)
ax.text(arc_radius + 0.4, 0.2, r'$\theta$', fontsize=16, va='center')

# --- 6. Indicação da Condição D >> a (Centralizado na linha de cota de D) ---
ax.annotate('', xy=(0, y_min_esquema + 0.5), xytext=(D, y_min_esquema + 0.5),
            arrowprops=dict(arrowstyle='<->', color='black', lw=1))

# Texto "D >> a" posicionado diretamente sobre a linha de cota
ax.text(D/2, y_min_esquema + 0.7, r'$D \gg a$', fontsize=16, ha='center', fontweight='bold')

# --- Ajustes Finais de Exibição ---
ax.set_xlim(-2, D + 3)
ax.set_ylim(y_min_esquema - 0.5, y_max_esquema + 0.5)
ax.set_aspect('equal') # Mantém a proporção geométrica perfeita no novo espaço de tela
ax.axis('off')         # Remove os eixos cartesianos padrão

plt.title('Esquema de Difração de Fraunhofer', fontsize=14, pad=15)
plt.tight_layout()

# Exibe a figura na tela
plt.show()
```

Para cada ponto da fenda, temos um feixe que vai até determinado ponto $P$ no anteparo, que faz um ângulo $\theta$ com a horizontal.
Podemos representar a onda vinda de cada ponto na representação fasorial complexa introduzida anteriormente. 
Como todas as ondas se originam da mesma frente de onda, a fase inicial pode ser escolhida como sendo zero. 
Também temos que todas tem a mesma polarização, então consideramos apenas a função escalar $\psi (\mathbf r, t)$. 
Além disso, como $\omega \sim 10^{15}$, detectamos apenas uma média temporal.
Portanto vamos desconsiderar termo $e^{-i\omega t}$.
Por fim, o índice de refração é constante, cada feixe se propaga em linha reta, de modo que $\mathbf k \cdot \mathbf r = kx$, onde x é a distância media ao longo da direção dos feixes.

Ou seja, basta apenas analisar $\psi= A \hat\psi$, onde $\hat\psi = e^{ikx}$ é o fasor da onda. 

Considere feixes partido de $N+1$ pontos equidistantes da fenda, de modo que a distância entre pontos adjacentes seja $a/N \equiv \Delta a$.
Considerando a difração de Fraunhofer ($D>>a$), os feixes são paralelos. 
Geometricamente, temos então que a diferença de caminho óptico é 
$$
L = \Delta a \sin\theta \;.
$$

Se o número de onda é $k$, a diferença de fase entre dois feixes adjacentes ao chegar em $P$ é então
$$
\Delta \phi = kL = k \Delta a\sin\theta \;.
$$

Para um número finito $N$, podemos então representar a onda resultante como sendo a soma dos $N+1$ fasores. 
A intensidade resultante é então proporcional ao quadrado do fasor resultante $\hat\psi_R$.

Utilizar geometria para encontrar o comprimento do fasor resultante pode parecer uma tarefa formidável, mas a situação fica mais simplies ao fazer $N \to \infty$.
Com $N$ arbitrariamente grande, temos que a linha poligonal tende a um arco de um círculo, e o fasor resultante é uma corda dessa circunferência. 
Temos
\begin{align*}
\Delta \phi = k \Delta a \sin\theta
\longrightarrow d\phi = k\sin\theta \;da 
\implies \phi = ka\sin\theta \;,
\end{align*}

onde $\phi$ é o ângulo entre os fasores das extremidades da fenda. 
Por outro lado, estes são justamente os vetores tangentes nas extremidades da curva. 
Portanto $\phi$ é o ângulo entre os raios ligados às extremidades da curva. 
Ou seja, $ka\sin\theta$ é a abertura ângular desse arco de circunferência.

Para encontrar o comprimento do arco, voltemos ao caso finito com 5 feixes.
Nesse caso, a intensidade total atravessando a fenda (sem interferência) é proporcional à $I_0 = (5A)^2$. 
Mas $5A$ é justamente o comprimento da liha poligonal formada plelos 5 fasores.
Como a $I_0$ é uma constante, devemos ter que a medida que $N$ cresce, $A$ diminui de modo que sempre temos $I_0 = [NA(N)]^2$. Portanto, fazendo $N \to \infty$, encontramos que o comprimento do arco é $\sqrt{I_0}$.

Dado o comprimento do arco e sua abertura angular, temos que o raio da circunferência é então 
$$
R = \frac{\sqrt{I_0}}{ka\sin\theta} \;.
$$
Pelo diagrama, temos que o comprimento do fasor resultante é 
\begin{gather*}
|\hat\psi_R| = 2 R \sin(\phi/2) = 2 \frac{\sqrt{I_0} \sin(ka\sin\theta/2)}{ka\sin\theta} = \sqrt{I_0} \operatorname{sinc}\left(\frac{ka\sin\theta}{2}\right) \;.
\end{gather*}

Assim, a intensidade resultante é proporcional a
$$
I(\theta) = I_0 \operatorname{sinc}^2\left(\frac{ka\sin\theta}{2}\right) \;.
$$

Essa função tem um máximo em $\theta =1$, sendo igual a $I_0$, e seus mínimos ocorrem quando $\theta = n\pi$, $n\in \mathbb Z^*$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vetor de pontos
x = np.linspace(-20, 20, 2000)

# sinc(x) = sin(x)/x
# Em NumPy, np.sinc(x) = sin(pi*x)/(pi*x)
# Portanto, para obter sin(x)/x usamos:
sinc = np.sinc(x / np.pi)

# sinc²(x)
y = sinc**2

# Gráfico
plt.figure(figsize=(8, 4.5))
plt.plot(x, y, color='navy', linewidth=2)

# Ticks em múltiplos de pi
n = np.arange(-6, 7)          # de -6π até 6π
xticks = n * np.pi

labels = []
for k in n:
    if k == 0:
        labels.append(r'$0$')
    elif k == 1:
        labels.append(r'$\pi$')
    elif k == -1:
        labels.append(r'$-\pi$')
    else:
        labels.append(rf'${k}\pi$')

plt.xticks(xticks, labels, fontsize=12)

plt.xlabel(r'$x$', fontsize=14)
plt.ylabel(r'$\mathrm{sinc}^2(x)$', fontsize=14)
plt.title(r'Gráfico da função $\mathrm{sinc}^2(x)$', fontsize=15)

plt.grid(True, linestyle='--', alpha=0.6)
plt.xlim(-20, 20)
plt.ylim(-0.02, 1.05)

plt.tight_layout()
plt.show()
```
