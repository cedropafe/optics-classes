# Equações de Maxwell
Na segunda metade do século XIX, James Clerk Maxwell compilou os resultados mais relevantes
de trabalhos acerca do eletromagnetismo, e formulou matematicamente a teoria eletromagnética.
Nesse trabalho, exploramos as equações de Maxwell em sua formulação integral e diferencial, assim
como mostramos que suas soluções assumem a forma de uma equação de onda.

## Lei de Gauss (eletricidade)
A lei de Gauss elétrica relaciona o campo elétrico com a distribuição de carga. Para o campo elétrico $\mathbf{E}$, essa lei afirma que o fluxo de $\mathbf{E}$ através de uma superfície fechada orientada (gaussiana) $S$ é proporcional à carga total localizada em seu interior, $Q$:
\begin{equation*}
\iint_{S} \mathbf{E} \cdot d\mathbf{A} = Q / \epsilon_0 \;,
\end{equation*} 
onde $\epsilon_0$ é a permissividade elétrica do vácuo. Considere uma carga pontual $q$ e uma esfera de raio $r$ centrada na carga. O campo devido à carga é perpendicular à superfície da esfera em todos os seus pontos, e seu módulo depende apenas de $r$, de modo que $\mathbf{E} \cdot d\mathbf{A}$ é uma constante. Aplicando na eq. acima e isolando $\mathbf{E}$, obtemos a expressão para o campo elétrico devido a uma carga pontual, que concorda com a expressão obtida através da lei de Coulomb:
\begin{equation*}
\mathbf{E} = \frac{1}{4\pi \epsilon_0} \frac{q}{r^2} \; . 
\end{equation*}

Fazendo uso do teorema da divergência, podemos reescrever a lei de Gauss elétrica como
\begin{equation*}
\frac{Q}{\epsilon_0}
= \iint_S \mathbf{E} \cdot d\mathbf{A}
= \iiint_V \nabla \cdot \mathbf{E} \,dV \; .
\end{equation*}

Se $\rho$ é a densidade volumétrica de carga, temos que $Q = \iiint_V \rho \,dV$. Substituindo na equação acima, obtemos a forma diferencial da lei de Gauss elétrica, também conhecida como equação de Poisson:
\begin{equation*}
\nabla \cdot \mathbf{E} = \rho / \epsilon_0 \; .
\end{equation*}

## Lei de Gauss (magnetismo)
Essa lei representa matematicamente o fato de não existirem monopolos magnéticos na natureza. A unidade básica do magnetismo é o dipolo magnético, que contém um polo norte (N) e um polo sul (S). A consequência disso para a configuração do campo magnético $\mathbf{B}$ é que não existe nenhum ponto no espaço que aja como fonte ou sorvedouro de $\mathbf{B}$. Pela definiçao de divergência de um campo vetorial, isso pode ser representado matematicamente por
\begin{equation*}
\nabla \cdot \mathbf{B} = 0 \; .
\end{equation*}

Considere uma superfície gaussiana $S$ que delimita um volume $V$, integrando $\nabla \cdot \mathbf{B}$ em $V$ e utilizando o teorema da divergência, obtemos
\begin{equation*}
\iiint_V \nabla \cdot \mathbf{B} \, dV = \iint_S \mathbf{B} \cdot d\mathbf{A} \; .
\end{equation*}
Mas temos que $\nabla \cdot \mathbf B = 0$. Assim, obtemos a forma integral da lei de Gauss magnética:
\begin{equation*}
\iint_S \mathbf B \cdot d\mathbf{A} = 0 \; ,
\end{equation*}

que nos diz que o fluxo do campo magnético é nulo através de qualquer superfície gaussiana. Com efeito, se o campo magnético é gerado por uma distribuição espacial de dipolos magnéticos, o fluxo de $\mathbf{B}$ através de $S$ devido a cada dipolo é nulo, pois qualquer linha de força vinda de N que atravessa a superfície em uma orientação, deve atravessar de novo, mas com orientação oposta, para ir para S. Portanto, o fluxo total é zero. 

## Lei de Ampère-Maxwell
Essa lei relaciona a distribuição de corrente elétrica e a variação de campo elétrico com o campo magnético.
Ela afirma que a circulação do campo magnético sobre uma curva fechada $C$ é igual à corrente elétrica que atravessa uma superfície orientada $S$ delimitada por $C$. Essa preposição pode ser representada matematicamente por
\begin{equation*}
    \int_C \mathbf B \cdot d\mathbf l = \mu_0 \iint_S \mathbf j \cdot d\mathbf A \; ,
\end{equation*}

onde $\mathbf j$ é o vetor densidade de corrente. Note que há infinitas superfícies delimitadas pela mesma curva.
Ampère afirma que o resultado é independente da superfície escolhida. 

Utilizando o teorema do rotacional, temos que $\int_C \mathbf B \cdot d\mathbf l = \iint_S (\nabla \times \mathbf B) \cdot d\mathbf A$. Como a igualdade deve se manter para qualquer superfície $S$, a forma diferencial da lei de Ampère é
\begin{equation*}
    \nabla \times \mathbf B = \mu_0 \mathbf j \; .
\end{equation*}

Considere agora um circuito com um capacitor de placas paralelas.
Seja $C$ uma curva fechada em torno do fio condutor.
Se considerarmos uma superfície $S$ que passa pelo fio, podemos usar o valor da corrente elétrica na lei de Ampère para determinar o campo $\mathbf B \neq 0$.
No entanto, podemos considerar uma superfície $S'$ com mesmo contorno de $S$, mas que passa pelo vão entre as placas, de modo que $\iint_{S'} \mathbf j \cdot d\mathbf S' = 0$, contradizendo o resultado anterior.

A inconsistência também surge na forma diferencial:
Tomando o divergente da forma diferencial da lei de Ampère e utilizando o fato de que div (rot $\mathbf F$) = 0 para qualquer campo $\mathbf F$, obtemos
\begin{equation*}
    0 = \nabla \cdot (\nabla \times \mathbf B) = \nabla \cdot (\mu_0 \mathbf j) = \mu_0 \nabla \cdot \mathbf j \; .
\end{equation*}
No entanto, pela equação da continuidade, temos
\begin{equation*}
    \nabla \cdot \mathbf j = - \partial_t \rho \;,
\end{equation*}
o que é uma contradição. Maxwell percebeu essas inconsistências e propôs uma alteração na lei de Ampère:
Pela eq. de Poisson, temos
\begin{equation*}
    \partial_t \rho = \partial_t (\epsilon_0 \nabla \cdot \mathbf E ) = \nabla \cdot (\epsilon_0 \partial_t \mathbf E) \;,
\end{equation*}

Então, a equação eq. de continuidade será satisfeita se a lei de Ampère for reformulada como
\begin{equation*}
    \nabla \times \mathbf B = \mu_0 \mathbf j + \epsilon_0 \mu_0 \partial_t \mathbf E \;.
\end{equation*}

Utilizando o teorema do rotacional, temos também
\begin{equation*}
    \int_C \mathbf B \cdot d\mathbf l = \iint_S \left( \mu_0\mathbf j + \epsilon_0\mu_0\partial_t\mathbf E \right) \cdot d\mathbf A \;.
\end{equation*}



## Lei de Faraday
A lei da indução de Faraday relaciona o campo elétrico com a variação no campo magnético. Experiências mostram que ao aproximar ou afastar um imã de uma espira condutora, uma corrente elétrica surge nela. Seja $C$ um circuito fechado. Se $\Phi_B$ é o fluxo do campo magnético através de uma superfície $S$ delimitada por $C$, então temos
\begin{equation*}
\mathcal{E} = -\dot{\Phi}_B \; ,
\end{equation*}

onde $\mathcal{E}$ é a força eletromotriz equivalente à corrente gerada no condutor. A eq. acima é a lei de Lenz, e diz que variações no fluxo magnético através de um condutor geram correntes que se opõem a essa variação. Por exemplo, ao aproximar um imã de uma espira condutora, a corrente gerada irá, por sua vez, gerar um novo campo magnético, que causará uma força de repulsão no imã. Ao afastar o imã, o mesmo ocorre, mas a força é de atração. Em ambos os casos, a força gerada pela corrente se opõe a variação de fluxo magnético. 


Decorre da lei de Gauss magnética que o fluxo de $\mathbf{B}$ sobre qualquer superfície não fechada depende apenas de seu contorno. Portanto, a lei de Lenz é valida para qualquer superfície de contorno $C$. Como a f.e.m. é a circulação do campo elétrico sobre um circuito, temos que 
\begin{equation*}
\int_C \mathbf{E} \cdot d\mathbf{l} = -\frac{d}{dt} \iint_S \mathbf{B} \cdot d\mathbf{A}.
\end{equation*}

Utilizando, o teorema do rotacional, obtemos 
\begin{equation*}
\nabla \times \mathbf{E} = - \partial_t \mathbf{B}.
\end{equation*}

Embora a teoria tenha sido formulada no contexto de correntes elétricas num circuito, as equações mostradas nessa sessão são válidas mesmo na ausência de condutores. 


## Ondas eletromagnéticas
A lei de Faraday nos diz que a variação no campo magnético gera um campo elétrico, enquanto a lei de Ampère-Maxwell nos diz que variação no campo elétrico gera um campo magnético.
Portanto, um uma perturbação eletromagnética gera um pulso que se propaga pelo espaço.
Nessa sessão, será mostrado que essa propagação dos campos $\mathbf E$ e $\mathbf B$ obedece uma equação de onda da forma
\begin{equation*}
    v^2 \nabla^2 \mathbf F = \partial_t^2 \mathbf F \;,
\end{equation*}
onde $v$ é a velocidade de propação da onda.

Vamos considerar as equações de Maxwell no vácuo ($\rho = 0,\, \mathbf j = 0$).
Utilizando uma identidade do operador $\nabla$, temos
\begin{equation*}
    \nabla \times \nabla \times \mathbf E = \nabla (\nabla \cdot \mathbf E) - \nabla^2 \mathbf E = -\nabla^2 \mathbf E \;,
\end{equation*}

onde foi utilizada a eq. de Poisson com $\rho = 0$. Por outro lado, utilizando a lei de Faraday em conjunto com a de Ampère-Maxwell, obtemos
\begin{align*}
    \nabla \times \nabla \times \mathbf E &= \nabla \times (-\partial_t \mathbf B) = -\partial_t (\nabla \times \mathbf B) \notag \\
    &= -\partial_t (\epsilon_0\mu_0\partial_t \mathbf E) = -\epsilon_0\mu_0\partial_t^2 \mathbf E
\end{align*}

Juntando as duas eqs. acima, obtemos
\begin{equation*}
    \nabla^2 \mathbf E = \epsilon_0\mu_0\partial_t^2 \mathbf E \;.
\end{equation*}

Analogamente, para o campo magnético $\mathbf B$, obtemos
\begin{equation*}
    \nabla^2 \mathbf B = \epsilon_0\mu_0\partial_t^2 \mathbf B \;.
\end{equation*}

Assim, os dois campos possuem a forma de uma equação de onda.
Concluímos que as perturbações se propagam na forma de ondas eletromagnéticas, cuja velocidade é $c \equiv (\epsilon_0 \mu_0)^{-1/2} = 299792458$ m/s.


É possível mostrar que soluções das equações de Maxwell são da forma
\begin{equation*}
    \mathbf B = \nabla \times \mathbf A \;, \quad \mathbf E = -\nabla \phi - \partial_t \mathbf A \;,
\end{equation*}

onde $\phi$ e $\mathbf A$ são respectivamente o potencial escalar e o potencial vetor, que obedecem a relação
\begin{equation*}
    \nabla \cdot \mathbf A + \epsilon_0\mu_0\partial_t \phi /c = 0 \;,
\end{equation*}

conhecida como condição de Lorentz.
