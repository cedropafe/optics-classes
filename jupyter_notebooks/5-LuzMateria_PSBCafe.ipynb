# Interação radiação matéria
Consideramos agora como a propagação da luz muda ao se propagar dentro de um meio dielétrico. 


## Eqs. de Maxwell Macroscópicas
Como um material é constituido de $\sim 10^{23}$ átomos, as grandezas utilizadas para descrever o estado são médias. 
Seja

- $\rho$ a densidade volumétrica de carga ;
- $\mathbf P$ a densidade volumétrica de dipolos elétricos, chamada de polarização ;
- $\mathbf M$ a densidade volumétrica de dipolos magnéticos, chamada de magnetização ;
- $\mathbf J$ a densidade de fluxo de corrente elétrica, chamada de densidade de corrente ;
- $\mathbf D = \epsilon_0 \mathbf E + \mathbf P$ o deslocamento elétrico ;
- $\mathbf H = \mu_0(\mathbf B + \mathbf M)$ a indução magnética.

As equações de Maxwell então se tornam
\begin{align*}
\text{rot } \mathbf E &= -\partial_t \mathbf H \;, \\[1em]
\text{rot } \mathbf B &= \partial_t \mathbf D + \mathbf J \;, \\[1em]
\text{div } \mathbf D &= \rho \;, \\[1em]
\text{div } \mathbf H &= 0 \; .
\end{align*}

A resposta do meio à influência de um campo eletromagnético externo é dada por
\begin{align*}
\mathbf J &= \sigma \mathbf E \;, \\[1em]
\mathbf D &= \epsilon \mathbf E \;, \\[1em]
\mathbf H &= \mu \mathbf B \;, \\[1em]
\mathbf P &= \chi \epsilon_0 \mathbf E \;,
\end{align*}

onde $\sigma$ é a condutividade elétrica do meio, e $\chi = \epsilon / \epsilon_0 -1$.

## Eq. de onda 
Consideraremos um meio isotrópico, não magnético, eletricamente neuro, e dielétrico. Assim, $\mathbf M$ e $\rho$ são ambos zero.
As eqs. de Maxwell são então
\begin{align*}
\text{rot } \mathbf E &= -\mu_0\partial_t \mathbf B \;, \\[1em]
\text{rot } \mathbf B &= \epsilon_0\partial_t \mathbf E + \partial_t \mathbf P + \mathbf J \;, \\[1em]
\text{div } \mathbf E &= -\frac{1}{\epsilon_0} \text{div } \mathbf P\;, \\[1em]
\text{div } \mathbf H &= 0 \; .
\end{align*}

A equação de onda é obtida ao tomar o rot da primeira equação, derivar a segunda parcialmente no tempo, e eliminar $\mathbf B$. Assim, temos
$$
\text{rot}^2 \mathbf E+\frac{1}{c^2} \partial_t^2 \mathbf E = - \mu_0 \partial_t^2 \mathbf P- \mu_0 \partial_t \mathbf J \;.
$$

## Resposta do meio
Considere que a posição do elétrons, com relação ao núcleo do átomo, é determinada por um vetor $\mathbf r$. Consideremos também que as dimensões de um átomo são desprezíveis, de modo que não há variação espacial do campo dentro do átomo. 

Seja $N$ o número de elétrons em uma unidade de volume. A polarização é então 
$$
\mathbf P = -Ne\mathbf r \;.
$$

Supondo que o elétrons está ligado à sua posição de equilíbrio por uma força elástica de constante $k$, temos
$$
-e \mathbf E = k \mathbf r \;.
$$

Assim,
$$
\mathbf P = \frac{Ne^2}{k} \mathbf E \;.
$$

A equação de movimento do elétrons é então 
$$
m \ddot{\mathbf r} + m\gamma \mathbf{\dot r} + k \mathbf r = -e \mathbf E \;,
$$
onde $m\gamma \dot{\mathbf r}$ é uma força de dissipação.

Supondo uma solução da forma $e^{-i\omega t}$, a equação se torna
$$
(-m\omega^2 - i\omega m \gamma + k) \mathbf r = -e \mathbf E \;.
$$

Assim, a polarização se torna
$$
\mathbf P = \frac{Ne^2 \mathbf E}{-m \omega^2 -i\omega m \gamma + k} =\frac{Ne^2/m \mathbf E}{\omega_0^2 - \omega^2 -i\omega \gamma} \;,
$$
onde $\omega_0 = \sqrt{k/m}$.

Substituindo esse $\mathbf P$ na eq. de onda, e utilizando $\mathbf J =0$, obtemos
\begin{gather*}
\text{rot}^2 \mathbf E+\frac{1}{c^2} \partial_t^2 \mathbf E = - \mu_0 \partial_t^2 \frac{Ne^2/m \mathbf E}{\omega_0^2 - \omega^2 -i\omega \gamma} \\[1em]
\implies \text{rot}^2 \mathbf E = -\frac{1}{c^2} \left( 1 + \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \right) \partial_t^2 \mathbf E \;.
\end{gather*}

Por outro lado, temos $\text{rot}^2 \mathbf E = \text{grad div } \mathbf E - \nabla^2 \mathbf E \;.$ Mas também temos

\begin{gather*}
\text{div } \mathbf E = -\frac{1}{\epsilon_0} \text{div } \mathbf P = -\frac{1}{\epsilon_0} \text{div } \left( \frac{Ne^2}{k} \mathbf E \right)
\implies \left( 1 + \frac{Ne^2}{\epsilon_0 k} \right) \text{div } \mathbf E = 0 \\[2em]
\therefore \text{rot}^2 \mathbf E = - \nabla^2 \mathbf E \;.
\end{gather*}

Substituíndo na eq. de onda, obtemos finalmente
$$
\nabla^2 \mathbf E = \frac{1}{c^2} \left( 1 + \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \right) \partial_t^2 \mathbf E \;.
$$

Tentamos uma solução da forma $\mathbf E = \mathbf E_0 e^{i(\mathcal K x -\omega t)}$. Substituindo na equação, obtemos
\begin{gather*}
- \mathcal K^2 \mathbf E = -\frac{\omega^2}{c^2} \left( 1 + \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \right) \mathbf E \\[1em]
\implies \left(\mathcal K^2 - \frac{\omega^2}{c^2} \left(1 + \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \right) \right) \mathbf E = 0 \\[2em]
\therefore \mathcal K = \frac{\omega}{c} \left(1+ \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \right)^{1/2} \;. 
\end{gather*}

A quantidade dentro da raiz quadrada é um número complexo $\mathcal N = n + i\xi$, chamado de índice de refração complexo. Assim, 
$$
\mathcal K = \frac{\omega}{c} \mathcal N = \alpha + i \beta \;,
$$
onde $\alpha = (\omega/c)n$, e $\beta = (\omega/c)\xi$. Assim, temos
$$
\mathbf E = \mathbf E_0 e^{-\beta x} e^{i(\alpha x - \omega t)} \;.
$$

A exponencial imaginária indica que temos uma onda plana com velocidade de fase $u = \omega / \alpha = c/n$. 
Enquanto que a exponencial real indica que a amplitude da onda decai exponencialmente a medida que a luz penetra no material. 
Como a densidade de energia da onda em um dado ponto do espaço é proporcial à $|\mathbf E|^2$, a quantidade $2\beta$ é chamada de coeficiente de absorção do meio. 

### Determinação de $\mathcal N$
Por um lado, temos
\begin{align*}
\frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} &= \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \left( \frac{\omega_0^2 - \omega^2 + i\omega \gamma}{\omega_0^2 - \omega^2 + i\omega \gamma} \right) \\[1em]
&= \frac{Ne^2}{m \epsilon_0} \left( \frac{\omega_0^2 - \omega^2 + i\omega \gamma}{(\omega_0^2 - \omega^2)^2 + \omega^2 \gamma^2} \right) \;.
\end{align*}

Por outro, temos
$$
(n+i\xi)^2 = n^2 - \xi^2 +2in\xi \;.
$$

Portanto, pela definição de $\mathcal N$, temos
\begin{gather*}
\mathcal N^2 = (n+i\xi)^2 = \frac{Ne^2/m\epsilon_0}{\omega_0^2 - \omega^2 -i\omega \gamma} \\[1em]
\implies n^2 - \xi^2 +2in\xi = 1 + \frac{Ne^2}{m \epsilon_0} \left( \frac{\omega_0^2 - \omega^2 + i\omega \gamma}{(\omega_0^2 - \omega^2)^2 + \omega^2 \gamma^2} \right) \;.
\end{gather*}

Igualando as partes real e imaginária, obtemos 
$$
\left\{
\begin{aligned}
n^2 - \xi^2 &= 1 + \frac{Ne^2}{m \epsilon_0} \left( \frac{\omega_0^2 - \omega^2}{(\omega_0^2 - \omega^2)^2 + \omega^2 \gamma^2} \right) \equiv A \\[1em]
2n\xi &= \frac{Ne^2}{m \epsilon_0} \left( \frac{\omega \gamma}{(\omega_0^2 - \omega^2)^2 + \omega^2 \gamma^2} \right) \equiv B
\end{aligned}
\right.
$$
Determinamos assim os parâmetros ópticos como sendo 
$$
n = \sqrt{\frac{\sqrt{A^2 + B^2}+A}{2}} \;, \qquad \xi = \sqrt{\frac{\sqrt{A^2 + B^2}-A}{2}} \;.
$$

Nos gráficos abaixo, vemos a depencência que o índice de refração tem com a frequência da radiação incidente.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- parameters for visualization (numbers don't matter) ---
omega_0 = 10.0
gamma = 1.2
A = 0.8  # scaling for the toy model

# frequency grid (allow negative too)
omega = np.linspace(0.0, 35.0, 2000)

# toy complex-response model to generate n and xi-like quantities
D = (omega_0**2 - omega**2)**2 + (omega**2 * gamma**2)
real_part = 1.0 + A * (omega_0**2 - omega**2) / D
imag_part = A * (omega * gamma) / D

# n and xi from (n + i*xi)^2 = real_part + i imag_part
mag = np.sqrt(real_part**2 + imag_part**2)
n = np.sqrt((mag + real_part) / 2.0)
xi = np.sign(imag_part) * np.sqrt((mag - real_part) / 2.0)

# --- plotting ---
plt.rcParams.update({"figure.dpi": 120, "axes.grid": True, "grid.alpha": 0.25})

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(7, 7))

# dashed vertical line at omega_0; label only on bottom plot
ax1.axvline(omega_0, color="k", linestyle="--", linewidth=1.5)
ax2.axvline(omega_0, color="k", linestyle="--", linewidth=1.5, label=r'$\omega_0$')

# Top: n vs omega
ax1.plot(omega, n, color="C0", linewidth=2)
ax1.axhline(1.0, color="gray", linestyle=":", linewidth=1)
ax1.set_ylabel(r"$n = \text{Re}\,\mathcal{N}$")
ax1.set_xlabel("")
ax1.set_yticks([1.0])  # show only "1"
ax1.tick_params(axis="both", which="both", labelbottom=False, labelleft=True)

# Bottom: xi vs omega
ax2.plot(omega, xi, color="C1", linewidth=2)
ax2.axhline(0.0, color="gray", linestyle=":", linewidth=1)
ax2.set_ylabel(r"$\xi = \text{Im}\,\mathcal{N}$")
ax2.set_xlabel(r"$\omega$")
ax2.set_yticks([0.0])  # show only "0"
ax2.legend(loc="best", frameon=True)
ax2.tick_params(axis="both", which="both", labelbottom=False, labelleft=True)

plt.tight_layout()
plt.show()
```
