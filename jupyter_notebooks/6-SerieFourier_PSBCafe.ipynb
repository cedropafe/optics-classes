# Séries de Fourier
Uma função $f$ é dita periódica com período $T>0$ se

- Se $x$ pertence ao domínio de $f$, então $x+T$ também o faz, e
- $f(x) = f(x+T)$, $\forall x$ no domínio.
  
Segue dessa definição que, se $T$ é período de $f$, então $nT$, $n\in \mathbb{N}$ também o é.
O menor $T$ que é período de $f$ é dito período fundamental de $f$.
Uma função constante é periódica com período arbitrário, mas sem período fundamental.

Afirmamos que qualquer função períodica de período $2L>0$ pode ser escrita na forma
$$
f(x) = \frac{A_0}{2} + \sum_{k=1}^{\infty} \left[ A_k \cos(\pi kx/L) + B_k \sin(\pi kx/L) \right] \;.
$$
Essa expressão é a expansão de $f$ em série de Fourier.

## Ortogonalidade de $\sin$ e $\cos$
Dado um espaço vetorial $V$ com produto interno $\langle \cdot,\cdot \rangle: V\times V \mapsto \mathbb{R}$ ou $\mathbb{C},$ definido, $\psi, \phi \in V$ são ditos ortogonais se, e somente se, $\langle \psi, \phi \rangle=0$. 
Um conjunto de vetores é dito mutualmente ortogonal se cada par distinto de vetores for ortogonal.

O espaço de funções é um espaço vetorial. Para podermos discutir a ortogonalidade de funções, precisamos então definir um produto interno.
Definimos então o P.I. entre duas funções $f$, $g$ definidas em algum intervalo $(a,b)$ como sendo 
$$
\langle f, g \rangle = \int_a^b \bar f(x) g(x) dx \;,
$$
onde a barra denota conjugação complexa.
Vamos mostrar que o conjunto contendo $\sin(n\pi x/L)$ e $\cos(m\pi x/L)$, $n,m\in\mathbb N$ é mutualmente ortogonal.
Em particular, fazendo $a=-L$ e $b=L$, temos 
\begin{align*}
\langle \sin(n\pi x/L) , \cos(m\pi x/L) \rangle &= 0 \; ; \\[1em]
\langle \cos(n\pi x/L) , \cos(m\pi x/L) \rangle &= \delta_{mn}L \; ; \\[1em]
\langle \sin(n\pi x/L) , \sin(m\pi x/L) \rangle &= \delta_{mn}L \; .
\end{align*}

#### Prova da primeira equação: 
Seja $n \neq m$, temos
\begin{align*}
\int_{-L}^L \sin(n\pi x/L) \cos(m\pi x/L) dx = \frac{1}{2} \int_{-L}^{L} \left[ \sin((n+m)\pi x/L) + \sin((n-m)\pi x/L) \right] dx \\[1em]
= -\frac{L}{2} \left[ \frac{\cos((n+m)\pi x/L)}{(n+m)\pi} + \frac{\cos((n-m)\pi x/L)}{(n-m)\pi} \right]_{-L}^{L} =0 \;,
\end{align*}
pois cosseno é função par, então os termos se cancelam após substituir os limites superior e inferior.

Seja agora $n=m$. Fazemos a substituição $cos(n\pi x/L) = u$. Assim, 
\begin{align*}
\int_{-L}^L \sin(n\pi x/L) \cos(n\pi x/L) dx = \frac{L}{n\pi} \int_{\cos(-n\pi)}^{\cos(n\pi)} u du = 0\; ,
\end{align*}
Pois cosseno é par, e portanto os limites de integração são iguais.

#### Prova da segunda equação
Seja $n \neq m$, temos
\begin{align*}
\int_{-L}^L \cos(n\pi x/L) \cos(m\pi x/L) dx = \frac{1}{2} \int_{-L}^{L} \left[ \cos((n+m)\pi x/L) + \cos((n-m)\pi x/L) \right] dx \\[1em]
= \frac{L}{2} \left[ \frac{\sin((n+m)\pi x/L)}{(n+m)\pi} + \frac{\sin((n-m)\pi x/L)}{(n-m)\pi} \right]_{-L}^{L} = 0 \;,
\end{align*}
pois, ao aplicar os limites de integração, os argumentos das funções seno serão múltimos inteiros de $\pi$.

Seja agora $n=m$, temos
\begin{align*}
\int_{-L}^L \cos^2(n\pi x/L) dx = \frac{1}{2}\int_{-L}^L \left[ 1 + \cos(2n\pi x/L)\right] = \left[ \frac{x}{2} + \frac{L}{4n\pi} \sin(2n\pi x/L) \right]_{-L}^{L} = L \;.
\end{align*}

#### Prova da terceira equação
Seja $n \neq m$, temos
\begin{align*}
\int_{-L}^L \sin(n\pi x/L) \sin(m\pi x/L) dx = -\frac{1}{2} \int_{-L}^{L} \left[ \cos((n+m)\pi x/L) - \cos((n-m)\pi x/L) \right] dx \\[1em]
= -\frac{L}{2} \left[ \frac{\sin((n+m)\pi x/L)}{(n+m)\pi} - \frac{\sin((n-m)\pi x/L)}{(n-m)\pi} \right]_{-L}^{L} = 0 \;.
\end{align*}

Seja agora $n=m$, temos
\begin{align*}
\int_{-L}^L \sin^2(n\pi x/L) dx = \frac{1}{2}\int_{-L}^L \left[ 1 - \cos(2n\pi x/L)\right] = \left[ \frac{x}{2} - \frac{L}{4n\pi} \sin(2n\pi x/L) \right]_{-L}^{L} = L \;.
\end{align*}



## Coeficientes de Fourier
Vamos agora mostrar um procedimento para encontrar os coeficientes $A$'s  e $B$'s da série de Fourier, chamados de coeficientes de Fourier, fazendo uso da ortogonalidade das funções.
Novamente, temos
$$
f(x) = \frac{A_0}{2} + \sum_{k=1}^{\infty} \left[ A_k \cos(\pi kx/L) + B_k \sin(\pi kx/L) \right] \;.
$$

Integrando $f$ de $-L$ a $L$, obtemos
$$
\int_{-L}^{L} f(x) dx = \int_{-L}^{L} \frac{A_0}{2} dx + \int_{-L}^{L} \sum_{k=1}^{\infty} \left[ A_k \cos(\pi kx/L) + B_k \sin(\pi kx/L) \right] dx = A_0 L \;,
$$
pois integrar qualquer função periódica por um périodo resulta em zero. Assim, 
$$
A_0 = \frac{1}{L} \int_{-L}^{L} f(x) dx \;.
$$

Para determinar $A_n$, $n \neq 0$, multiplicamos $f(x)$ por $\cos(n\pi x/L)$, e então integramos em $x$:
\begin{align*}
\int_{-L}^{L} \cos(n\pi x/L) f(x) dx &= \frac{A_0}{2} \int_{-L}^{L} \cos(n\pi x/L) dx + \sum_{k=1}^{\infty} \left[ A_k \int_{-L}^{L} \cos(n\pi x/L) \cos(\pi kx/L) dx + B_k \int_{-L}^{L} \cos(n\pi x/L) \sin(\pi kx/L) dx \right] \\[1em]
&= 0 + \sum_{k=1}^{\infty} \left[ A_k \delta_{kn} L + 0 \right] = A_n L \;.
\end{align*}

Assim, temos
$$
A_n = \frac{1}{L} \int_{-L}^L \cos(n\pi x/L) f(x) dx \;.
$$
Note que a fórmula acima vale mesmo para $n=0$, já que o termo constante da expansão é $A_0/2$. e não $A_0$.

Para encontrar $B_n$, o procedimento é o mesmo, mas com $\sin(n\pi x/L)$:
$$
B_n = \frac{1}{L} \int_{-L}^L \sin(n\pi x/L) f(x) dx \;.
$$

Dado que o procedimento para encontrar os coeficientes pode ser aplicado para qualquer função periódica, consideramos essa demonstração como uma prova informal de que qualquer função periódica pode ser expandida em série de Fourier.

Também é possível simplificar o procedimento analisando a paridade da função $f$:

- Se $f$ é par, então $B_n=0$, $\forall n\in\mathbb N$ ;
- Se $f$ é ímpar, então $A_n=0$, $\forall n \in \mathbb N_0$ .
